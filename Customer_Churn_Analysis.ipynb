```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales.csv")

df.head()
```

```python
total_customers = df["Customer Name"].nunique()

print("Total Customers:", total_customers)
```

```python
repeat_customers = df["Customer Name"].value_counts()

repeat_customers = repeat_customers[repeat_customers > 1]

print("Repeat Customers:")
print(repeat_customers.head(10))
```

```python
top_customers = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_customers)
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum()

print(segment_sales)
```

```python
segment_sales.plot(kind="pie", autopct='%1.1f%%')

plt.title("Customer Segment Distribution")

plt.show()
```

```python
region_customers = df.groupby("Region")["Customer ID"].nunique()

print(region_customers)
```

```python
region_customers.plot(kind="bar")

plt.title("Customers by Region")
plt.xlabel("Region")
plt.ylabel("Number of Customers")

plt.show()
```

```python
customer_orders = df["Customer Name"].value_counts()

churn_customers = customer_orders[customer_orders == 1]

print("Possible Churn Customers:")
print(churn_customers.head(10))
```

# Business Insights

1. Some customers purchase repeatedly and contribute high revenue.

2. Customers with single orders may indicate churn risk.

3. Consumer segment generates significant sales.

4. Regional customer distribution helps improve marketing strategy.

# Conclusion

This analysis identified repeat customers, churn-risk customers, and customer segment trends.
Businesses can use these insights to improve retention and customer engagement.
