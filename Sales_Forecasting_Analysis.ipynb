```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales.csv")

df.head()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()

print(monthly_sales)
```

```python
monthly_sales.plot(kind="line", marker="o", figsize=(10,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="bar")

plt.title("Regional Sales Performance")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="pie", autopct='%1.1f%%')

plt.title("Category Contribution")

plt.show()
```

```python
best_month = monthly_sales.idxmax()

print("Highest Sales Month:", best_month)
```

# Forecast Insights

1. Sales show strong seasonal patterns.

2. Some months generate significantly higher revenue.

3. Technology and high-performing regions drive business growth.

4. Forecast analysis helps businesses prepare inventory and marketing strategies.

# Conclusion

This analysis identified monthly sales trends and business growth patterns.
Organizations can use forecasting insights for strategic planning and future sales optimization.

```python
rolling_sales = monthly_sales.rolling(window=2).mean()

rolling_sales.plot(kind="line", marker="o")

plt.title("Moving Average Sales Trend")
plt.xlabel("Month")
plt.ylabel("Average Sales")

plt.show()
```
